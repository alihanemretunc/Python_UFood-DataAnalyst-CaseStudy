```python
age_groups = [(23, 30), (31, 40), (41,50), (51, 60), (61, 70), (71, 85)]

def assign_age_group(Age):
    for age_range in age_groups:
        if age_range[0] <= Age <= age_range[1]:
            return f"{age_range[0]} - {age_range[1]}"
    return "Unknown"

food['Age_Group'] = food['Age'].apply(assign_age_group)
```

### 1. Age group 30-70 were spending more money, but less likely to accept campaigns. Higher volume here though.

### 2. Catalog was more likely to accept campaigns, but in person spends more. Recommend a split between all.
- 40% catalog, 30% store, 30% web

### 3. Focus on people with no kids (or less kids).

### 4. Education - no impact - don't target any group

### 5. Marital Status - doesn't play a big part. Married, Single, Together -- they spend more money.
